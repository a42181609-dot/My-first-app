```python
class Cow:
  def __init__(self,cow,breed,country):
    self.cow = cow
    self.breed = breed
    self.country = country

  def display_info(self):
    print(f"{self.cow} {self.breed} {self.country}")

cow1 = Cow("cow", "frezen", "from astrelia")
cow1.display_info()
```
