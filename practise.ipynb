```python
import requests
url = "https://httpbin.org/status/200"

def check_status(url, label):
    try:
        import requests
        "https://httpbin.org/status/200"
        r = requests.get(url, timeout=8)
        code = r.status_code
        if code == 200:
            status = "SUCCESS"
        elif code == 401:
            status = "UNAUTHORIZED  bad API key"
        elif code == 404:
            status = "NOT FOUND  city or endpoint does't exist"
        elif code == 429:
            status = "RATE LIMITED  slow down your requests"
        else:
            status = f"OTHER  {code}"
        print(f"[{code}] {label}: {status}")
    except requests.exceptions.ConnectionError:
        print(f"[ERR] {label}: Timeout")

print("Testing diffrerent API response:")
print("=" * 55)


# 404 city not found
check_status("https://httpbin.org/status/404", "City not found")






```
